```python
import torch
import torch.nn as nn
import torch.nn.functional as F

###############################################
# Basic Conv Block
###############################################

class ConvBNAct(nn.Module):
    def __init__(self, c1, c2, k=3, s=1, p=1):
        super().__init__()
        self.conv = nn.Conv2d(c1, c2, k, s, p, bias=False)
        self.bn = nn.BatchNorm2d(c2)
        self.act = nn.SiLU()

    def forward(self, x):
        return self.act(self.bn(self.conv(x)))


###############################################
# CSP Block (YOLO style)
###############################################

class CSPBlock(nn.Module):
    def __init__(self, c):
        super().__init__()
        self.conv1 = ConvBNAct(c, c)
        self.conv2 = ConvBNAct(c, c)

    def forward(self, x):
        return x + self.conv2(self.conv1(x))


###############################################
# Micro YOLO Stem
###############################################

class MicroYOLOStem(nn.Module):
    """
    Designed for 28x28 OCTMNIST images
    """

    def __init__(self):
        super().__init__()

        self.conv1 = ConvBNAct(1, 32, 3, 1, 1)
        self.conv2 = ConvBNAct(32, 32, 3, 1, 1)
        self.csp = CSPBlock(32)

    def forward(self, x):

        x = self.conv1(x)
        x = self.conv2(x)
        x = self.csp(x)

        return x


###############################################
# YOLO Backbone
###############################################

class YOLOBackbone(nn.Module):

    def __init__(self):

        super().__init__()

        self.stage1 = ConvBNAct(32, 64, 3, 2, 1)
        self.csp1 = CSPBlock(64)

        self.stage2 = ConvBNAct(64, 128, 3, 2, 1)
        self.csp2 = CSPBlock(128)

    def forward(self, x):

        f1 = x

        x = self.stage1(x)
        x = self.csp1(x)

        f2 = x

        x = self.stage2(x)
        x = self.csp2(x)

        f3 = x

        return f1, f2, f3


###############################################
# Lightweight FPN
###############################################

class LightFPN(nn.Module):

    def __init__(self):
        super().__init__()

        self.up = nn.Upsample(scale_factor=2)

        self.conv1 = ConvBNAct(128, 64, 1, 1, 0)
        self.conv2 = ConvBNAct(128, 64, 3, 1, 1)

    def forward(self, f1, f2, f3):

        x = self.up(self.conv1(f3))

        x = torch.cat([x, f2], dim=1)

        x = self.conv2(x)

        return x


###############################################
# SAM Prior Module
###############################################

class SAMPriorModule(nn.Module):

    def __init__(self):

        super().__init__()

        self.conv1 = nn.Conv2d(1, 32, 3, padding=1)
        self.conv2 = nn.Conv2d(32, 1, 1)

    def forward(self, x):

        mask = F.relu(self.conv1(x))
        mask = torch.sigmoid(self.conv2(mask))

        return mask


###############################################
# Cross Mask Fusion
###############################################

class CrossMaskFusion(nn.Module):

    def __init__(self, c):

        super().__init__()

        self.q = nn.Conv2d(c, c, 1)
        self.k = nn.Conv2d(1, c, 1)
        self.v = nn.Conv2d(c, c, 1)

        self.out = nn.Conv2d(c, c, 1)

    def forward(self, feature, mask):

        Q = self.q(feature)

        K = self.k(mask)

        V = self.v(feature)

        attn = torch.sigmoid(Q * K)

        fused = attn * V

        out = self.out(fused)

        return out + feature


###############################################
# Structural Tokenizer
###############################################

class StructuralTokenizer(nn.Module):

    def __init__(self, in_channels, token_dim=256):

        super().__init__()

        self.pool = nn.AdaptiveAvgPool2d(1)

        self.proj = nn.Linear(in_channels, token_dim)

    def forward(self, feature):

        b, c, h, w = feature.shape

        pooled = self.pool(feature).view(b, c)

        token = self.proj(pooled)

        return token


###############################################
# Simplified Mamba Block
###############################################

class MambaBlock(nn.Module):

    def __init__(self, dim):

        super().__init__()

        self.norm = nn.LayerNorm(dim)

        self.linear1 = nn.Linear(dim, dim)

        self.linear2 = nn.Linear(dim, dim)

    def forward(self, x):

        residual = x

        x = self.norm(x)

        x = F.relu(self.linear1(x))

        x = self.linear2(x)

        return x + residual


###############################################
# Mamba Sequence Module
###############################################

class MambaSequence(nn.Module):

    def __init__(self, dim=256, depth=2):

        super().__init__()

        self.layers = nn.ModuleList([
            MambaBlock(dim) for _ in range(depth)
        ])

    def forward(self, x):

        for layer in self.layers:
            x = layer(x)

        return x


###############################################
# Structure Pooling
###############################################

class StructurePooling(nn.Module):

    def __init__(self, dim):

        super().__init__()

        self.weight = nn.Linear(dim, 1)

    def forward(self, tokens):

        w = torch.softmax(self.weight(tokens), dim=1)

        pooled = torch.sum(w * tokens, dim=1)

        return pooled


###############################################
# Classifier
###############################################

class Classifier(nn.Module):

    def __init__(self, dim, num_classes):

        super().__init__()

        self.fc1 = nn.Linear(dim, dim)

        self.fc2 = nn.Linear(dim, num_classes)

    def forward(self, x):

        x = F.relu(self.fc1(x))

        return self.fc2(x)


###############################################
# Full YOLO-SAM-Mamba Network
###############################################

class YOLOSAMMambaNet(nn.Module):

    def __init__(self, num_classes=4):

        super().__init__()

        self.stem = MicroYOLOStem()

        self.backbone = YOLOBackbone()

        self.fpn = LightFPN()

        self.sam_prior = SAMPriorModule()

        self.fusion = CrossMaskFusion(64)

        self.tokenizer = StructuralTokenizer(64)

        self.mamba = MambaSequence()

        self.pool = StructurePooling(256)

        self.classifier = Classifier(256, num_classes)

    def forward(self, x):

        x = self.stem(x)

        f1, f2, f3 = self.backbone(x)

        fpn_feature = self.fpn(f1, f2, f3)

        mask = self.sam_prior(x)

        fused = self.fusion(fpn_feature, mask)

        token = self.tokenizer(fused)

        token = token.unsqueeze(1)

        seq = self.mamba(token)

        pooled = self.pool(seq)

        logits = self.classifier(pooled)

        return logits
```

```python
import torch
import torch.nn.functional as F # Import F for interpolation
import types

# Original model instantiation
model = YOLOSAMMambaNet()

def fixed_forward_impl(self, x_input):
    original_x = x_input # Store the original input

    x = self.stem(x_input)

    f1, f2, f3 = self.backbone(x)

    fpn_feature = self.fpn(f1, f2, f3)

    mask = self.sam_prior(original_x)

    # Resize mask to match fpn_feature's spatial dimensions (14x14)
    # fpn_feature has shape (B, C, 14, 14), mask has (B, 1, 28, 28)
    mask_resized = F.interpolate(mask, size=fpn_feature.shape[2:], mode='bilinear', align_corners=False)

    fused = self.fusion(fpn_feature, mask_resized)

    token = self.tokenizer(fused)

    token = token.unsqueeze(1)

    seq = self.mamba(token)

    pooled = self.pool(seq)

    logits = self.classifier(pooled)

    return logits

# Patch the forward method of the model instance
model.forward = types.MethodType(fixed_forward_impl, model)

x = torch.randn(8,1,28,28)

y = model(x)

print(y.shape)
```

```python
import torch
import torchvision.transforms.functional as TF
import random


class OCTAugmentation:

    def __init__(self):

        pass

    def __call__(self, x):

        if random.random() < 0.5:
            x = TF.hflip(x)

        if random.random() < 0.5:
            x = TF.vflip(x)

        if random.random() < 0.3:
            x = TF.rotate(x, angle=random.uniform(-15,15))

        if random.random() < 0.3:
            noise = torch.randn_like(x) * 0.02
            x = x + noise

        x = torch.clamp(x,0,1)

        return x
```

```python
import torch
from sklearn.metrics import accuracy_score, roc_auc_score

def compute_metrics(logits, labels):

    probs = torch.softmax(logits, dim=1)

    preds = torch.argmax(probs, dim=1)

    acc = accuracy_score(labels.cpu(), preds.cpu())

    try:

        auc = roc_auc_score(labels.cpu(), probs.cpu(), multi_class="ovr")

    except:

        auc = 0

    return acc, auc
```

```python
import torch
from torch.cuda.amp import autocast, GradScaler
from tqdm import tqdm

class Trainer:

    def __init__(self, model, optimizer, criterion, device):

        self.model = model
        self.optimizer = optimizer
        self.criterion = criterion
        self.device = device

        self.scaler = GradScaler()

    def train_epoch(self, loader):

        self.model.train()

        total_loss = 0

        for img, label in tqdm(loader):

            img = img.to(self.device)
            # Squeeze labels to handle (B, 1) vs (B) shape issues
            label = label.to(self.device).squeeze().long()

            self.optimizer.zero_grad()

            with autocast():

                logits = self.model(img)

                loss = self.criterion(logits, label)

            self.scaler.scale(loss).backward()

            self.scaler.step(self.optimizer)

            self.scaler.update()

            total_loss += loss.item()

        return total_loss / len(loader)
```

```python
import torch

class Evaluator:

    def __init__(self, model, device):

        self.model = model
        self.device = device

    def evaluate(self, loader):

        self.model.eval()

        logits_all = []
        labels_all = []

        with torch.no_grad():

            for img, label in loader:

                img = img.to(self.device)

                logits = self.model(img)

                logits_all.append(logits.cpu())
                # Squeeze labels to (B) from (B, 1)
                labels_all.append(label.squeeze().long())

        logits_all = torch.cat(logits_all)
        labels_all = torch.cat(labels_all)

        # compute_metrics is already defined in the global scope
        acc, auc = compute_metrics(logits_all, labels_all)

        return acc, auc
```
